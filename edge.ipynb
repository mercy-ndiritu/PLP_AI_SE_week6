```python
# 1. Import TensorFlow
import tensorflow as tf

# 2. Load your trained Keras model
# (replace with your actual model path or variable)
model = tf.keras.models.load_model("models/recycle_mobilenetv2.h5")

# 3. Convert to TensorFlow Lite
converter = tf.lite.TFLiteConverter.from_keras_model(model)
tflite_model = converter.convert()

# 4. Save the TFLite model
with open("model.tflite", "wb") as f:
    f.write(tflite_model)

print("✅ Conversion complete! Saved as model.tflite")
```
